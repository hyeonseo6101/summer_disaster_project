```python
code = pd.read_excel('C:/Users/hyeon/OneDrive/Desktop/2026-1/공모전/데이터분석/데이터/행정구역/법정동코드 조회자료.xlsx',
                     dtype = {'법정동코드' : str})
code['법정동명'] = code['법정동명'].str.strip()
```

```python
final_df = pd.merge(df, code[['법정동명', '법정동코드']], on='법정동명', how='left')
```

```python
# 1. 법정동별 전체 주거용 건물 수 집계
total_counts = final_df.groupby('법정동코드').size().reset_index(name='전체_건물수')

# 2. 법정동별 반지하 건물 수 집계 (지하층수가 1 이상인 건물만 필터링)
basement_df = final_df[(final_df['지하층수'] != 0.0) & (final_df['지하층수'] != -1.0)]
basement_counts = basement_df.groupby('법정동코드').size().reset_index(name='반지하_개수')
```

```python
# 3. 전체 건물 수 데이터와 반지하 건물 수 데이터 병합
analysis_df = pd.merge(total_counts, basement_counts, on='법정동코드', how='left')

# 4. 반지하 건물이 없는 동은 NaN이 되므로 0으로 채우기
analysis_df['반지하_개수'] = analysis_df['반지하_개수'].fillna(0)

# 5. 비율(%) 또는 소수점 비율 계산
analysis_df['반지하_비율'] = analysis_df['반지하_개수'] / analysis_df['전체_건물수']
```

```python
# 8자리 키 생성 및 병합
gdf['merge_key'] = gdf['EMD_CD'].astype(str).str[:8]
analysis_df['merge_key'] = analysis_df['법정동코드'].astype(str).str[:8]

merged_gdf = gdf.merge(analysis_df, on='merge_key', how='left')
```

```python
# 카카오 API 좌표계로 변환 후 중심점 추출
merged_gdf_4326 = merged_gdf.to_crs(epsg=4326)
merged_gdf_4326['centroid'] = merged_gdf_4326.geometry.centroid
merged_gdf_4326['lon'] = merged_gdf_4326['centroid'].x
merged_gdf_4326['lat'] = merged_gdf_4326['centroid'].y
```

```python
KAKAO_API_KEY = "21c049207110a2d9d78be20602686acd"
```

```python
import time

def get_admin_dong(lon, lat):
    url = f"https://dapi.kakao.com/v2/local/geo/coord2regioncode.json?x={lon}&y={lat}"
    headers = {"Authorization": f"KakaoAK {KAKAO_API_KEY}"}
    
    try:
        response = requests.get(url, headers=headers)
        
        time.sleep(0.1) 
        
        if response.status_code == 200:
            documents = response.json().get('documents', [])
            for doc in documents:
                if doc['region_type'] == 'H':  # 'H' : 행정동
                    return doc['region_3depth_name']
        else:
            print(f"API Error: {response.status_code}") 
            
    except Exception as e:
        print(f"Request Error: {e}")
        return None
        
    return None
```

```python
# API 호출을 통해 각 법정동 폴리곤에 행정동 이름 추가
merged_gdf_4326['행정동명'] = merged_gdf_4326.progress_apply(
    lambda row: get_admin_dong(row['lon'], row['lat']), axis=1
)
```

```python
# API 오류 등으로 매핑되지 않은 곳 예외 처리
merged_gdf_4326['행정동명'] = merged_gdf_4326['행정동명'].fillna('미분류')
```

```python
# 행정동 기준으로 경계 병합
admin_gdf = merged_gdf_4326.dissolve(by='행정동명', aggfunc={'전체_건물수': 'sum', '반지하_개수': 'sum'}).reset_index()

# 행정동 기준 반지하 비율 다시 계산
admin_gdf['반지하_비율'] = admin_gdf['반지하_개수'] / admin_gdf['전체_건물수']
admin_gdf['반지하_비율'] = admin_gdf['반지하_비율'].fillna(0)
```

```python
# 시각화 실행
m = admin_gdf.explore(
    column='반지하_비율',        
    cmap='YlOrRd',                       
    tooltip=['행정동명', '전체_건물수', '반지하_개수', '반지하_비율'], 
    style_kwds={'weight': 0.5, 'color': 'gray', 'fillOpacity': 0.8}
)
```

```python
### S1 지표 추출

# 시군구코드명에서 자치구만 추출
# 결측치나 타입 오류를 방지하기 위해 str 변환 후 분리
final_df['자치구'] = final_df['시군구코드명'].apply(lambda x: str(x).split()[-1])
```

```python
# 자치구별 전체 건물 수 산출
gu_total_counts = final_df.groupby('자치구').size().reset_index(name='전체_건물수')
```

```python
# 자치구별 반지하 건물 수 산출 (지하층수가 1 이상인 경우)
gu_basement_df = final_df[(final_df['지하층수'] != 0.0) & (final_df['지하층수'] != -1.0)]
gu_basement_counts = gu_basement_df.groupby('자치구').size().reset_index(name='반지하_개수')
```

```python
# 데이터 병합 및 S1 원시값(반지하 비율) 계산
s1_df = pd.merge(gu_total_counts, gu_basement_counts, on='자치구', how='left')
```

```python
# 반지하 건물이 없는 자치구의 경우 NaN을 0으로 처리
s1_df['반지하_개수'] = s1_df['반지하_개수'].fillna(0)
s1_df['S1_원시값'] = s1_df['반지하_개수'] / s1_df['전체_건물수']
```

```python
# Min-Max 정규화 (0 ~ 1 범위로 변환)
# 수식: (x - min) / (max - min)
s1_min = s1_df['S1_원시값'].min()
s1_max = s1_df['S1_원시값'].max()

s1_df['S1_정규화'] = (s1_df['S1_원시값'] - s1_min) / (s1_max - s1_min)
```

```python
s1_df_sorted = s1_df.sort_values(by='S1_정규화', ascending=False).reset_index(drop=True)
s1_df_sorted.head()
```

    - 건물 수 집계 : 지하층수가 1 이상인 반지하 건물수와 전체 주거용 건물 수

### 2. 반지하 비율 산출
- 전체 주거 환경 대비 얼마나 밀집해있는지
    - 비율 계산 : 반지하 건물 수 / 전체 주거용 건물 수

### 3. 취약성 평가 지수
- 정규화 과정 포함
    - MIN-MAX 정규화 : 반지하 밀집도가 가장 높은 구는 1, 가장 낮은 구는 0
    - S1 : 얼마나 위험에 노출되기 쉬운 상태인지 나타내는 민감도 지표 S1으로 사용
